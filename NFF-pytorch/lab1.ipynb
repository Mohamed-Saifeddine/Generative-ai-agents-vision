```python
x = torch.tensor(3.0 , requires_grad=True)
w = torch.tensor(4.0 , requires_grad=True)
b = torch.tensor(6.0 , requires_grad=True)
y= w * x + b
loss = (y - 10) ** 2
```

```python
y
```

```python
loss
```

```python
loss.backward()
```

```python
print(w.grad.item())
```

```python
print(b.grad.item())
```

```python
dw, db= torch.autograd.grad(loss, (w , b))
```
